```python

import sys
import os
import platform
import importlib.util


def get_python_info():
    """获取 Python 版本"""
    print("=" * 50)
    print("1. Python 版本")
    print("=" * 50)
    print(f"Python 版本：{platform.python_version()}")
    print(f"Python 完整版本：{sys.version.split()[0]}")
    print(f"Python 可执行文件：{sys.executable}")


def get_pytorch_info():
    """获取 PyTorch 版本及计算设备"""
    print("\n" + "=" * 50)
    print("2. PyTorch 版本")
    print("=" * 50)

    if importlib.util.find_spec("torch") is None:
        print("PyTorch：未安装")
        print("计算设备：无法检测（请先安装 PyTorch）")
        return

    try:
        import torch

        print(f"PyTorch 版本：{torch.__version__}")
        print(f"CUDA 是否可用：{torch.cuda.is_available()}")

        if torch.cuda.is_available():
            print("计算设备：GPU")
            print(f"GPU 数量：{torch.cuda.device_count()}")

            for i in range(torch.cuda.device_count()):
                print(f"GPU {i}：{torch.cuda.get_device_name(i)}")

            print(f"当前 CUDA 设备编号：{torch.cuda.current_device()}")
        else:
            print("计算设备：CPU")
            if hasattr(torch.backends, "mps"):
                if torch.backends.mps.is_available():
                    print("提示：当前 PyTorch 还检测到 Apple MPS 可用")
    except Exception as e:
        print(f"PyTorch 检测失败：{e}")


def get_environment():
    """识别 Google Colab、Kaggle 或本地环境"""
    print("\n" + "=" * 50)
    print("3. 当前运行环境")
    print("=" * 50)

    import os
    import sys

    # Kaggle 的主要特征
    kaggle_markers = [
        os.path.isdir("/kaggle"),
        os.path.isdir("/kaggle/working"),
        os.path.isdir("/kaggle/input"),
        "KAGGLE_KERNEL_RUN_TYPE" in os.environ,
        "KAGGLE_URL_BASE" in os.environ,
    ]

    # Google Colab 的主要特征
    colab_markers = [
        "COLAB_RELEASE_TAG" in os.environ,
        "COLAB_BACKEND_VERSION" in os.environ,
        "COLAB_GPU" in os.environ,
        "COLAB_JUPYTER_IP" in os.environ,
    ]

    is_kaggle = any(kaggle_markers)
    is_colab = any(colab_markers)

    # 优先使用平台专属标志；冲突时检查运行时目录
    if is_kaggle and not is_colab:
        environment = "Kaggle"
    elif is_colab and not is_kaggle:
        environment = "Google Colab"
    elif is_kaggle and is_colab:
        if os.path.isdir("/kaggle/working"):
            environment = "Kaggle"
        else:
            environment = "Google Colab"
    else:
        try:
            import google.colab
            environment = "Google Colab（根据模块推断）"
        except ImportError:
            environment = "本地环境"

    print(f"运行环境：{environment}")
    print(f"Python 路径：{sys.executable}")
    print(f"当前目录：{os.getcwd()}")
    print(f"操作系统：{platform.system()} {platform.release()}")
    print(f"Kaggle 标志：{is_kaggle}")
    print(f"Colab 标志：{is_colab}")

def main():
    print("\n        Python 环境检测工具")
    get_python_info()
    get_pytorch_info()
    get_environment()
    print("\n检测完成！")


if __name__ == "__main__":
    main()
```
