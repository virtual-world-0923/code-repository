```python
import os
import re
import zipfile

# ================= 配置区域 =================
# 1. 在这里粘贴你的 Google Drive 共享链接
url = "https://drive.google.com/file/d/1MS36CtVzH-ofJygprIcbo7Nob7k7Fo_o/view?usp=sharing"

# 2. 设置下载后的文件名（可以保持默认）
output_zip = "downloaded_file.zip"
# ============================================

# 使用正则表达式从链接中精准提取文件 ID
match = re.search(r'/d/([^/]+)', url)

if match:
    file_id = match.group(1)
    print(f"✅ 成功提取文件 ID: {file_id}")

    print("⏳ 开始下载文件，请稍候...")
    # 使用 --id 语法，可以自动跳过大文件的 Google 病毒扫描警告
    os.system(f'gdown --id {file_id} -O {output_zip}')

    # 检查文件是否成功下载
    if os.path.exists(output_zip):
        print("⚡ 下载完成！开始进行【防乱码】安全解压...")

        # ==================== ✨ 核心修复：Python 逆向解码去乱码 ====================
        with zipfile.ZipFile(output_zip, "r") as zf:
            for member in zf.infolist():
                try:
                    # 针对 Windows 经典压缩包：先用 cp437 还原字节，再用 gbk 解码
                    filename = member.filename.encode('cp437').decode('gbk')
                except:
                    try:
                        # 针对 Mac/Linux 压缩包：尝试用 utf-8 解码
                        filename = member.filename.encode('cp437').decode('utf-8')
                    except:
                        filename = member.filename

                # 统一路径斜杠，防止 Linux 环境识别错误
                filename = filename.replace('\\', '/')
                member.filename = filename

                # 解压单个文件到当前目录
                zf.extract(member, path="./")
        # ============================================================================

        print("🎉 解压完成！请刷新左侧文件浏览器，中文文件名已恢复正常。")
    else:
        print("❌ 下载失败，请检查文件是否确实开启了“任何拥有链接的人都可以查看”的权限。")

else:
    print("❌ 链接格式不正确，无法提取文件 ID。请确保使用的是类似 https://drive.google.com/file/d/... 的共享链接。")
```
